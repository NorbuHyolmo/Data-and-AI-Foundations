### Task 4: Operator Precedence

```python
print("Result:", 5 > 4 and 10 == 4)
print("First condition state True, Second condition states False") 
print("And operator states True only when both conditions are True")
```

```python
print("Result:",5 > 4 or 10 == 4)
print("OR operator states that even if one condition is true, the statement is true")
```

```python
print("Result:",not 5 > 10)
print("Condition is false but not false = true")
```
