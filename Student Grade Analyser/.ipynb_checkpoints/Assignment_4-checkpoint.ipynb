```python
import math
class Circle:
    shapes_created = 0

    def __init__(self, radius):
        if radius <= 0:
            raise ValueError(f"Radius must be positive, got {radius}")
        self.radius = radius
        Circle.shapes_created += 1

    def area(self):
        result = math.pi * self.radius * self.radius
        return round(result, 2)

    def perimeter(self):
        result = 2 * math.pi * self.radius
        return round(result, 2)

    @staticmethod
    def cm_to_inch(cm):
        return round(cm * 0.3937, 2)

    @staticmethod
    def is_valid_dimension(val):
        if val > 0:
            return True
        else:
            return False

    @classmethod
    def total_created(cls):
        return f"Circles created: {cls.shapes_created}"

    def __str__(self):
        return f"Circle(r={self.radius})"
```

```python
class Rectangle:
    shapes_created = 0

    def __init__(self, width, height):
        if width <= 0 or height <= 0:
            raise ValueError(f"Dimensions must be positive, got {width} and {height}")
        self.width  = width
        self.height = height
        Rectangle.shapes_created += 1

    def area(self):
        result = self.width * self.height
        return round(result, 2)

    def perimeter(self):
        result = 2 * (self.width + self.height)
        return round(result, 2)

    def is_square(self):
        if self.width == self.height:
            return True
        else:
            return False

    @staticmethod
    def cm_to_inch(cm):
        return round(cm * 0.3937, 2)

    @staticmethod
    def is_valid_dimension(val):
        if val > 0:
            return True
        else:
            return False

    @classmethod
    def total_created(cls):
        return f"Rectangles created: {cls.shapes_created}"

    def __str__(self):
        return f"Rectangle({self.width} x {self.height})"
```

```python
c1 = Circle(5)
c2 = Circle(3)
c3 = Circle(7)
r1 = Rectangle(4, 6)
r2 = Rectangle(5, 5)

print(c1)
print(f"Area      : {c1.area()}")
print(f"Perimeter : {c1.perimeter()}")

print(c2)
print(f"Area      : {c2.area()}")
print(f"Perimeter : {c2.perimeter()}")
```

```python
print(r1)
print(f"Area      : {r1.area()}")
print(f"Perimeter : {r1.perimeter()}")
print(f"Is Square : {r1.is_square()}")

print(r2)
print(f"Area      : {r2.area()}")
print(f"Perimeter : {r2.perimeter()}")
print(f"Is Square : {r2.is_square()}")
```

```python
print(f"cm to inch (10cm) : {Circle.cm_to_inch(10)}")
print(f"Is valid (-2)     : {Rectangle.is_valid_dimension(-2)}")
print(f"Is valid (5)      : {Rectangle.is_valid_dimension(5)}")
```

```python
print(Circle.total_created())
print(Rectangle.total_created())
```

```python
try:
    Circle(-3)
except ValueError as e:
    print(f"Caught ValueError → {e}")
```

```python
try:
    Rectangle(0, 5)
except ValueError as e:
    print(f"Caught ValueError → {e}")
```
